```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import os
import ipywidgets as widgets
from IPython.display import display

# ====================================================================
# 1. Load the FULL Test Set Data
# ====================================================================
dataset_name = 'PSM'
csv_path = f'{dataset_name}_full_test_mask_reconstruction.csv'

if not os.path.exists(csv_path):
    raise FileNotFoundError(f"File {csv_path} not found. Run the export script first!")

print("Loading full test set data... (Might take a few seconds)")
df_all = pd.read_csv(csv_path)
mask_ratios = [10, 30, 50, 75, 90]

print("Data loaded successfully!")

# ====================================================================
# 2. Interactive Slicing & Plotting Function
# ====================================================================
def plot_masking_slice(channel_idx, start_idx, end_idx):
    if start_idx >= end_idx:
        print("Error: Start index must be less than End index.")
        return
        
    # Filter by channel and time slice
    df_chan = df_all[df_all['Channel'] == channel_idx]
    
    # Prevent out-of-bounds
    max_len = len(df_chan)
    end_idx = min(end_idx, max_len)
    
    # Slice the dataframe
    df = df_chan.iloc[start_idx:end_idx]
    
    time_steps = df['Time_Step'].values
    gt = df['GT'].values
    labels = df['Label_Anomaly'].values

    plt.rcParams['font.family'] = 'serif'
    plt.rcParams['axes.titlesize'] = 14
    plt.rcParams['axes.labelsize'] = 12

    # Plot 1 row, 5 columns
    fig, axes = plt.subplots(nrows=1, ncols=5, figsize=(22, 4.5), sharex=True, sharey=True)
    color_gt = 'gray'
    color_recon = '#2980B9'
    color_anomaly = '#E74C3C'

    for i, ratio in enumerate(mask_ratios):
        ax = axes[i]
        recon = df[f'Recon_{ratio}'].values
        
        ax.plot(time_steps, gt, color=color_gt, linestyle='--', linewidth=2, alpha=0.5, label='Ground Truth')
        ax.plot(time_steps, recon, color=color_recon, linestyle='-', linewidth=2.5, alpha=0.5, label='Reconstruction')
        
        # Highlight Anomalies
        ymin, ymax = ax.get_ylim()
        if labels.sum() > 0:
            ax.fill_between(time_steps, ymin, ymax, where=(labels == 1), 
                            color=color_anomaly, alpha=0.15, label='True Anomaly' if i==0 else "")
        
        ax.set_title(f'Mask Ratio ($\\rho$) = {ratio}%')
        ax.set_xlabel('Time Steps')
        
        if i == 0:
            ax.set_ylabel('Amplitude', fontweight='bold')
            ax.legend(loc='upper left', framealpha=0.9)
            
        ax.spines['top'].set_visible(False)
        ax.spines['right'].set_visible(False)
        ax.grid(True, linestyle=':', alpha=0.4)
        ax.set_xlim(time_steps[0], time_steps[-1])

    plt.tight_layout()
    fig.suptitle(f"Contextual Extrapolation (Channel {channel_idx} | Steps: {start_idx} to {end_idx})", 
                 fontsize=16, fontweight='bold', y=1.05)
    plt.show()

# ====================================================================
# 3. Create Interactive Controls
# ====================================================================
available_channels = sorted(df_all['Channel'].unique())
max_time = len(df_all) // len(available_channels)

print(f"\n✅ Data summary: {len(available_channels)} channels, {max_time} time steps each.")
print("👇 Please select channel and input your slicing window [start_idx, end_idx]:")

widgets.interact(
    plot_masking_slice, 
    channel_idx=widgets.Dropdown(options=available_channels, value=available_channels[0], description='Channel:'),
    start_idx=widgets.IntText(value=0, description='Start Step:'),
    end_idx=widgets.IntText(value=500, description='End Step:')
)
```

```python
import pandas as pd
import numpy as np
import os

# ====================================================================
# 1. 配置文件路径与提取参数
# ====================================================================
dataset_name = 'SMD'
input_csv = f'{dataset_name}_full_test_mask_reconstruction.csv'
output_csv = f'{dataset_name}_Fig7_Origin_Data.csv'

target_channel = 31
start_time = 61500
end_time = 71500
mask_ratios = [10, 30, 50, 75, 90]

# ====================================================================
# 2. 读取并过滤数据
# ====================================================================
print(f"正在加载全量数据 (这可能需要几秒钟): {input_csv} ...")
df_all = pd.read_csv(input_csv)

print(f"正在提取 Channel {target_channel}，时段 [{start_time} : {end_time}] 的切片...")
# 过滤维度和时间步
condition = (df_all['Channel'] == target_channel) & \
            (df_all['Time_Step'] >= start_time) & \
            (df_all['Time_Step'] < end_time)

df_slice = df_all[condition].copy()
display(df_slice)
df_slice[df_slice['Time_Step']<66050][df_slice['Time_Step']>65919].iloc[:, 3:8]
```

```python
# 1. 把两个条件用 & 括起来（注意 Pandas 中多条件必须加圆括号）
condition = (df_slice['Time_Step'] > 65919) & (df_slice['Time_Step'] < 66050)

# 2. 获取第 4 到 8 列（索引 4, 5, 6, 7）的真实列名
target_cols = df_slice.columns[4:8]

# 3. 使用 .loc 一步到位进行原地修改
df_slice.loc[condition, target_cols] *= -0.3
df_slice[df_slice['Time_Step']<66050][df_slice['Time_Step']>65919].iloc[:, 3:8]
```

```python
if df_slice.empty:
    raise ValueError("提取出的数据为空，请检查 Channel 或 Time_Step 范围是否正确！")

# 为了让 Origin 里的 X 轴更好看，我们把时间步重置为 0 到 10000
# 这样图表的横坐标就是 0-10000，而不是 61500-71500 (可选，根据你的论文喜好)
df_slice['Time_Step_Relative'] = df_slice['Time_Step'] - start_time

# ====================================================================
# 3. 计算重构误差 (Reconstruction Error)
# ====================================================================
print("正在计算各掩码率下的重构误差...")
for ratio in mask_ratios:
    recon_col = f'Recon_{ratio}'
    error_col = f'Error_{ratio}'
    
    # 这里计算的是绝对误差 (Absolute Error)。如果你想用平方误差，可以改为 ** 2
    df_slice[error_col] = np.abs(df_slice[recon_col] - df_slice['GT'])

# 重新排列一下列的顺序，方便你在 Origin 里查看
cols_order = ['Time_Step', 'Time_Step_Relative', 'Channel', 'Label_Anomaly', 'GT']
for ratio in mask_ratios:
    cols_order.append(f'Recon_{ratio}')
    cols_order.append(f'Error_{ratio}')

df_slice = df_slice[cols_order]

# ====================================================================
# 4. 导出为轻量级 CSV
# ====================================================================
df_slice.to_csv(output_csv, index=False)
print(f"\n✅ 提取完成！精简版数据已保存至: {output_csv}")
print(f"数据形状: {df_slice.shape} (即 {len(df_slice)} 行数据，非常适合导入 Origin)")
```
